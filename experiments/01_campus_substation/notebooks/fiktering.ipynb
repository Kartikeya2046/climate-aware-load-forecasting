```python
import pandas as pd

# Load dataset
df = pd.read_csv("../data/interim/hourly_energy.csv")

# Remove rows where energy > 400 kWh
df_filtered = df[df["energy_kwh.1"] <= 400]

# (Optional) reset index after removing rows
df_filtered = df_filtered.reset_index(drop=True)

# Save the cleaned dataset
df_filtered.to_csv("../data/interim/hourly_energy_filtered.csv", index=False)

# Check result
print(df_filtered.head())
print("Remaining rows:", len(df_filtered))
```

```python
import pandas as pd

# Load the dataset
df = pd.read_csv("../data/interim/hourly_energy.csv")

# Keep only valid readings (0 to 400 kWh)
df = df[(df["energy_kwh.1"] >= 0) & (df["energy_kwh.1"] <= 400)]

# Reset index (optional but cleaner)
df = df.reset_index(drop=True)

# Save back to the same file (overwrite)
df.to_csv("../data/interim/hourly_energy.csv", index=False)

print("Cleaning complete. File updated successfully.")
print("Remaining rows:", len(df))
```

```python
print("Min energy:", df["energy_kwh.1"].min())
print("Max energy:", df["energy_kwh.1"].max())
```

```python
import pandas as pd

# Load both datasets
energy_df = pd.read_csv("../data/interim/hourly_energy.csv")
weather_df = pd.read_csv("../data/interim/weather_hourly.csv")

# Convert timestamp columns to datetime (important for proper matching)
energy_df["timestamp"] = pd.to_datetime(energy_df["timestamp"])
weather_df["timestamp"] = pd.to_datetime(weather_df["timestamp"])

# Keep only timestamps present in BOTH datasets
merged_df = pd.merge(energy_df, weather_df, on="timestamp", how="inner")

# Sort by timestamp (optional but good practice)
merged_df = merged_df.sort_values("timestamp")

# Reset index
merged_df = merged_df.reset_index(drop=True)

# Save final dataset for model training
merged_df.to_csv("../data/processed/final_training_dataset.csv", index=False)

print("Final dataset created successfully.")
print("Number of rows in final dataset:", len(merged_df))
```
