```python
import pandas as pd

# load file
df = pd.read_csv("../data/interim/SUB_S_clean_withoutgrouping.csv")

# convert to datetime (important)
df["cleaned bucket"] = pd.to_datetime(df["cleaned bucket"])

# group by bucket and sum energy
hourly_sum = (
    df.groupby("cleaned bucket", as_index=False)["energy_kwh.1"]
    .sum()
)

# sort by time
hourly_sum = hourly_sum.sort_values("cleaned bucket")

print(hourly_sum)
```

```python
hourly_sum.to_csv("../data/interim/hourly_energy.csv", index=False)
```

```python
import pandas as pd

df = pd.read_csv("../data/interim/cleaned_data.csv")

df["timestamp"] = pd.to_datetime(df["timestamp"])

df["hour_bucket"] = df["timestamp"].dt.floor("h")

df_hourly = (
    df.groupby("hour_bucket")
    .mean(numeric_only=True)
    .reset_index()
)

df_hourly = df_hourly.rename(columns={"hour_bucket": "timestamp"})

df_hourly = df_hourly.sort_values("timestamp")

df_hourly.to_csv("../data/interim/weather_hourly.csv", index=False)
```
