```python
import pandas as pd

df = pd.read_csv("../data/interim/cleaned_weather_DUQ.csv")

# Clean column names
df.columns = df.columns.str.strip()

# Use your correct column name
time_col = 'datetime'   # change if needed

# Convert to string (important for replace operations)
df[time_col] = df[time_col].astype(str)

# 🔥 Step 1: Handle "24:00:00"
mask = df[time_col].str.contains("24:00:00")

# Replace 24:00:00 → 00:00:00
df.loc[mask, time_col] = df.loc[mask, time_col].str.replace("24:00:00", "00:00:00")

# Convert to datetime
df[time_col] = pd.to_datetime(df[time_col], format='%Y-%m-%d %H:%M:%S')

# 🔥 Step 2: Add +1 day where it was originally 24:00:00
df.loc[mask, time_col] = df.loc[mask, time_col] + pd.Timedelta(days=1)

# 🔥 Step 3: Shift years by +25
def shift_year(dt):
    try:
        return dt.replace(year=dt.year + 25)
    except ValueError:
        return dt.replace(month=2, day=28, year=dt.year + 25)

df[time_col] = df[time_col].apply(shift_year)

# Sort
df = df.sort_values(by=time_col).reset_index(drop=True)

# Save
df.to_csv("../data/processed/weather_DUQ_2006-2016.csv", index=False)

print("✅ Fixed 24:00 issue + shifted years successfully.")
```

```python
import pandas as pd

# Load your dataset
df = pd.read_csv("../data/processed/FINAL_MERGED_DAY_LOAD_2006_2016.csv")  # change if needed

# Clean column names
df.columns = df.columns.str.strip()

# Set your datetime column name
time_col = 'datetime'   # change if needed

# Convert to datetime (if not already)
df[time_col] = pd.to_datetime(df[time_col])

# 🔥 Remove rows where year is 2010 or 2017
df = df[~df[time_col].dt.year.isin([2010, 2017,2016])]

# Sort (good practice)
df = df.sort_values(by=time_col).reset_index(drop=True)

# Save cleaned data
df.to_csv("../data/processed/FINAL_MERGED_DAY_LOAD_2006_2016.csv", index=False)

print("✅ Rows from years 2010 and 2017 and 2016 removed successfully.")
```
