```python
import pandas as pd

# Load CSV
df = pd.read_csv("../data/interim/kentucky_weather_1962_1980.csv")

# Convert to datetime
df['datetime'] = pd.to_datetime(df['datetime'])

# Shift year by +43 (1962 → 2005)
df['datetime'] = df['datetime'] + pd.DateOffset(years=43)

# Save updated file
df.to_csv("../data/processed/kentucky_weather_2005_2018.csv", index=False)

print("Done ✅")
```

```python
import pandas as pd

df = pd.read_csv("../data/raw/Indiana_consumption_2005_18.csv")

# Clean column names
df.columns = df.columns.str.strip().str.lower()

# Now safely use 'datetime'
df['datetime'] = pd.to_datetime(df['datetime'])

# Remove 2004
df = df[df['datetime'].dt.year != 2004]

df.to_csv("../data/interim/cleaned.csv", index=False)

print("✅ Done")
```

```python
import pandas as pd

# Load your dataset
df = pd.read_csv("../data/raw/kentucky_weather_reconstructed.csv")

# Convert to datetime (VERY IMPORTANT)
df["datetime"] = pd.to_datetime(df["datetime"])

# Filter between 1962 and 1980 (inclusive)
df_filtered = df[
    (df["datetime"] >= "1962-01-01") &
    (df["datetime"] <= "1980-12-31 23:59:59")
]

# Sort just in case
df_filtered = df_filtered.sort_values("datetime")

# Reset index
df_filtered = df_filtered.reset_index(drop=True)

# Save new CSV
df_filtered.to_csv("../data/interim/kentucky_weather_1962_1980.csv", index=False)

print("✅ Done! Filtered file saved as kentucky_weather_1962_1980.csv")
```
